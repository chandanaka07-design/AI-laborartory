```python
# ==========================================
# Measures of Central Tendency - IRIS Dataset
# ==========================================

import pandas as pd
from sklearn.datasets import load_iris
from scipy import stats
from scipy.stats import gmean, hmean

# ------------------------------------------
# Step 1: Load the IRIS Dataset
# ------------------------------------------

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("First Five Records")
print(df.head())

# ------------------------------------------
# Step 2: Mean
# ------------------------------------------

print("\nMean")
print(df.mean())

# ------------------------------------------
# Step 3: Median
# ------------------------------------------

print("\nMedian")
print(df.median())

# ------------------------------------------
# Step 4: Mode
# ------------------------------------------

print("\nMode")
print(df.mode())

# ------------------------------------------
# Step 5: Geometric Mean
# ------------------------------------------

print("\nGeometric Mean")

for column in df.columns:
    print(column, ":", gmean(df[column]))

    # ------------------------------------------
    # Step 6: Harmonic Mean
    # ------------------------------------------

    print("\nHarmonic Mean")

    for column in df.columns:
        print(column, ":", hmean(df[column]))
```
